## Defining a Qualifying Purchase

For this analysis, I define a qualifying purchase as an order with an
`order_status` of `delivered`.

I use delivered orders because the project focuses on completed customer
purchasing behavior. Orders in intermediate states such as `shipped`,
`processing`, and `invoiced`, as well as `canceled` and `unavailable` orders,
are excluded from the purchase definition.

This definition will be applied consistently when calculating customer
purchase frequency, monetary behavior, and the future repeat-purchase target.

```python
qual_orders_per_cust = customer_orders[customer_orders["order_status"] == "delivered"].groupby("customer_unique_id")["order_id"].nunique()
qual_orders_per_cust.shape[0]

qual_orders_per_cust.value_counts().sort_index()
```

# Choosing a temporal split 

```python
orders_by_month = (qualifying_orders.set_index("order_purchase_timestamp").resample("ME").size())

orders_by_month
```

```python
import matplotlib.pyplot as plt

a = orders_by_month.plot(
    kind="bar",
    figsize=(12, 5),
    title="Delivered Orders by Month"
)

a.set_xlabel("Month")
a.set_ylabel("Orders")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The prediction cutoff was set to March 31, 2018. Customer features are calculated from qualifying purchases occurring on or before the cutoff,

 while the repeat-purchase target is determined from qualifying purchases between April 1 and September 30, 2018.
 
  This provides a sufficiently long historical period for customer level feature construction.
  
  As well as a complete six month future observation period while excluding the incomplete October 2018 data.

```python
#making date reference points 

cutoff_date = pd.Timestamp("2018-03-31")

future_start = pd.Timestamp("2018-04-01") 

future_end = pd.Timestamp("2018-09-30")
```

```python
#checking if the data types are correct for timestamp

qualifying_orders.dtypes

```

```python
#seperating delivered orders (qualifying orders)

historical_orders = qualifying_orders[qualifying_orders["order_purchase_timestamp"] <= cutoff_date].copy()  #making it a seperate dataframe with copy

future_orders = qualifying_orders[(qualifying_orders["order_purchase_timestamp"] >= future_start) &
                                  (qualifying_orders["order_purchase_timestamp"] <= future_end)].copy() #making this a seperate dataframe also

#crosscheck before continuing


print("Historical orders:", len(historical_orders))
print("Future orders:", len(future_orders))

print("Latest historical purchase:",historical_orders["order_purchase_timestamp"].max())

print("Earliest future purchase:",future_orders["order_purchase_timestamp"].min())

print("Latest future purchase:",future_orders["order_purchase_timestamp"].max())
```
