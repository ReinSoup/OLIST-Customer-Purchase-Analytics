## Defining a Qualifying Purchase

For this analysis, I define a qualifying purchase as an order with an
`order_status` of `delivered`.

I use delivered orders because the project focuses on completed customer
purchasing behavior. Orders in intermediate states such as `shipped`,
`processing`, and `invoiced`, as well as `canceled` and `unavailable` orders,
are excluded from the purchase definition.

This definition will be applied consistently when calculating customer
purchase frequency, monetary behavior, and the future repeat-purchase target.

## Building Customer-Level RFM Features

The historical orders dataframe is currently at the order grain, where one row represents one delivered order.

To create RFM features, I will aggregate the historical order data to the customer grain, where one row represents one `customer_unique_id`.

The features will be calculated as follows:

- Recency: days since the customer's most recent historical purchase
- Frequency: number of unique historical delivered orders
- Monetary: total historical order value
- Average Order Value: Monetary divided by Frequency

```python
last_purchase = historical_orders.groupby("customer_unique_id")["order_purchase_timestamp"].max().reset_index()
last_purchase.head()

#.max() to capture last purchase
```

customer unique id > has multiple orders made > groupby puts all orders belonging to same customer together

.max() selects the latest purchase for that unique id. hence we get last purchase.

reset index, otherwise customer id would become index

``order_purchase_timestamp`` displays properly with accurate timestamp values rather than `00:00:00`

## Calculate Recency

recency = cutoff date - last purchase date

```python
print(cutoff_date)
```

```python
#add recency col to last purchase table

last_purchase["recency"] = (cutoff_date - last_purchase["order_purchase_timestamp"])
```

```python
last_purchase.head()
```

when performing date - date = pandas gives a timedelta value

date1 - date2 = 123days 00:00:00

for the model we want a numeric feature representing days.

### Convert Recency to Days

```python
last_purchase["recency"] = last_purchase["recency"].dt.total_seconds() / (24 * 60 * 60)
```
