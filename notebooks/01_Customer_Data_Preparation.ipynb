```python
customer_orders = orders.merge(customers[["customer_id", 'customer_unique_id']], on = "customer_id", how = "left") 
customer_orders.head()
```

```python
print("orders: ", len(orders))
print("customer orders merged: ", len(customer_orders))

print("orders without customer unique id: ", customer_orders["customer_unique_id"].isna().sum())
```

```python
customer_order_counts = customer_orders.groupby("customer_unique_id")["order_id"].nunique()
customer_order_counts.value_counts().sort_index()
```

```python
customer_purchase_dates = (customer_orders.groupby("customer_unique_id")["order_purchase_timestamp"].agg(["min", "max", "count"]).reset_index())
customer_purchase_dates = customer_purchase_dates.rename(columns = {

    "min" : "First Purchase Date",
    "max" : "Last Purchase Date",
    "count": "Order Count"
    })
customer_purchase_dates.describe()
```
